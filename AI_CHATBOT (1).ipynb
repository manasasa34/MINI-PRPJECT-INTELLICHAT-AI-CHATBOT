```python
from google import genai
from google.colab import userdata

# Get API key securely
api_key = userdata.get("GEMINI_API_KEY")

# Create Gemini client
client = genai.Client(api_key=api_key)

print("Gemini connected successfully!")
```

```python
response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents="Hello! Introduce yourself as an AI assistant."
)

print(response.text)
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents="Hello! Introduce yourself as an AI assistant."
)

print(response.text)
```

```python
chat = client.chats.create(
    model="gemini-3.5-flash-lite"
)

print("IntelliChat is ready!")

while True:
    question = input("You: ")

    if question.lower() == "exit":
        print("Goodbye!")
        break

    try:
        response = chat.send_message(question)
        print("IntelliChat:", response.text)

    except Exception as e:
        print("Temporary error:", e)
```

```python
import gradio as gr

# Function to process user messages
def chatbot_response(message, history):

    try:
        response = chat.send_message(message)
        return response.text

    except Exception as e:
        return f"Sorry, something went wrong: {e}"


# Create the chatbot interface
demo = gr.ChatInterface(
    fn=chatbot_response,
    title="🤖 IntelliChat AI Assistant",
    description="Your personal AI assistant powered by Google Gemini",
    textbox=gr.Textbox(
        placeholder="Ask me anything...",
        container=False
    )
)

# Launch the application
demo.launch(share=True)
```
