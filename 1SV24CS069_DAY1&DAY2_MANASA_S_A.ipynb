## 4. Introduction to Data Visualisation

**Data visualisation** means representing data using charts and graphs so that patterns, comparisons, trends and unusual values are easier to understand.

Why visualise?
- A chart can reveal a pattern that is difficult to notice in raw rows.
- It helps compare products, regions and customer segments.
- It makes trends and unusual observations easier to communicate.

### Choosing a chart
| Question | Suitable chart |
|---|---|
| How does a value change over time? | Line |
| Which category is highest? | Bar |
| How are values distributed? | Histogram |
| Are two numeric variables related? | Scatter |
| What share does each small group contribute? | Pie |
| Are there outliers? | Box plot |

## 10. Correlation analysis
**Correlation** measures the strength and direction of a linear relationship between two numeric variables. Its coefficient ranges from **-1 to +1**.

- **Positive correlation:** as one variable increases, the other tends to increase.
- **Negative correlation:** as one variable increases, the other tends to decrease.
- **Near zero correlation:** little or no linear relationship is observed.

Correlation does not prove causation.

### 10.1 Select numeric columns

```python
numeric_df = df.select_dtypes(include="number")
numeric_df.head()
```

### 10.2 Calculate the correlation matrix

### How to explain the heatmap
- Values close to **+1** indicate a strong positive linear association.
- Values close to **-1** indicate a strong negative linear association.
- Values close to **0** indicate a weak linear association.
- The diagonal is always 1 because each variable is perfectly correlated with itself.

A correlation coefficient does not tell us that one variable causes another. Other variables or business conditions may influence both.
