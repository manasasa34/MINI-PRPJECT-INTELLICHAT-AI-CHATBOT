```python
#CNN – Handwritten Digit Recognition
#Objective: Build a CNN model to recognize handwritten digits from 0 to 9 using the MNIST dataset.
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D
from tensorflow.keras.layers import Flatten, Dense, Dropout
```

```python
#load dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Training data:", x_train.shape)
print("Testing data:", x_test.shape)

print("Training labels:", y_train.shape)
print("Testing labels:", y_test.shape)
```

```python
#Data preprocessing
x_train = x_train / 255.0
x_test = x_test / 255.0

# Add channel dimension
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

print(x_train.shape)
print(x_test.shape)
```

```python
#build CNN Model
model = Sequential()

# Convolution layer
model.add(Conv2D(32, (3,3), activation='relu',
                 input_shape=(28,28,1)))

# Pooling layer
model.add(MaxPooling2D(pool_size=(2,2)))

# Second convolution layer
model.add(Conv2D(64, (3,3), activation='relu'))

# Second pooling layer
model.add(MaxPooling2D(pool_size=(2,2)))

# Convert 2D feature maps into 1D
model.add(Flatten())

# Fully connected layer
model.add(Dense(64, activation='relu'))

# Dropout to reduce overfitting
model.add(Dropout(0.3))

# Output layer
model.add(Dense(10, activation='softmax'))
```

```python
#Display model architecture
model.summary()
```

```python
#compile mdoel
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
#train cnn model
history = model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.1
)
```

```python
#evaluate mdoel
loss, accuracy = model.evaluate(x_test, y_test)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

```python
#RNN-Recurrent Neural Network
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
```

```python
#load IMDB dataset
vocab_size = 10000

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.imdb.load_data(
    num_words=vocab_size
)

print("Training data:", x_train.shape)
print("Testing data:", x_test.shape)

print("First review:", x_train[0])
print("Sentiment:", y_train[0])
```

```python
#Convert word IDs into readable words
word_index = tf.keras.datasets.imdb.get_word_index()

reverse_word_index = {
    index + 3: word for word, index in word_index.items()
}

reverse_word_index[0] = "<PAD>"
reverse_word_index[1] = "<START>"
reverse_word_index[2] = "<UNK>"
reverse_word_index[3] = "<UNUSED>"

review = " ".join(
    reverse_word_index.get(i, "?") for i in x_train[0]
)

print(review[:1000])
```

```python
#Data preprocessing – Padding
#Every review has a different number of words. We make all reviews the same length.
max_length = 200

x_train = tf.keras.utils.pad_sequences(
    x_train,
    maxlen=max_length,
    padding="post",
    truncating="post"
)

x_test = tf.keras.utils.pad_sequences(
    x_test,
    maxlen=max_length,
    padding="post",
    truncating="post"
)

print(x_train.shape)
print(x_test.shape)
```

```python
#build rnn model
model = Sequential()

model.add(Embedding(10000, 64))

model.add(SimpleRNN(64))

model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
#compile mdoel
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
#train model
history = model.fit(
    x_train[:12000],
    y_train[:12000],
    epochs=3,
    batch_size=128,
    validation_split=0.2
)
```

```python
#evaluate model
loss, accuracy = model.evaluate(
    x_test[:5000],
    y_test[:5000]
)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

```python
#GRU-Gated Recurrent Unit
#Objective: Build a GRU model to classify movie reviews as positive or negative.
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GRU, Dense
```

```python
#load dataset
vocab_size = 10000

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.imdb.load_data(
    num_words=vocab_size
)

print("Training data:", x_train.shape)
print("Testing data:", x_test.shape)

print("First review:", x_train[0])
print("Sentiment:", y_train[0])
```

```python
#Data preprocessing – Padding
#Every review has a different length. We convert all reviews into 200 tokens.max_length = 200

x_train = tf.keras.utils.pad_sequences(
    x_train,
    maxlen=max_length,
    padding='post',
    truncating='post'
)

x_test = tf.keras.utils.pad_sequences(
    x_test,
    maxlen=max_length,
    padding='post',
    truncating='post'
)

print("Training shape:", x_train.shape)
print("Testing shape:", x_test.shape)
```

```python
#build GRU model
model = Sequential()

# Convert word IDs into vectors
model.add(Embedding(10000, 64))

# GRU layer
model.add(GRU(64))

# Output layer
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
#compile model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
#Evalaute model
loss, accuracy = model.evaluate(
    x_test[:5000],
    y_test[:5000]
)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```
